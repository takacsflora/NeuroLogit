The repository contains tools to analyse neural data [here](https://github.com/takacsflora/NeuroLogit/tree/main/NeuroLogit/src/ephys). Including
- Combined conditions choice/stimulus/action probability (`cccp`), see [Steinmetz et al. 2019](https://www.nature.com/articles/s41586-019-1787-x) for further details.
- Decoding choice using adding neural predictors to decode choice (`decodng_choice`)
- Encoding models of average responses, i.e. firing rate in a single time bin dependent on stimuli/choices (`enoding_avg`)
- Kernel regression models i.e. firing rate aligned to distinct events in time bins (`encoding_kernel`)

Preprocessed data is available upon request until formal publication of my paper.
